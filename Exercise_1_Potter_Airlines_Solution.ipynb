## Part C — Pricing function


```python
def calculate_fare(base_fare, days_until_departure, seats_remaining, capacity, route_popularity):
    # These two checks are provided.
    if capacity <= 0:
        raise ValueError("capacity must be greater than 0")
    if seats_remaining < 0 or seats_remaining > capacity:
        raise ValueError("seats_remaining must be between 0 and capacity")

    # SOLUTION 1
    if days_until_departure <= 7:
        time_factor = 1.35
    if 8 <=days_until_departure <= 21:
        time_factor = 1.10
    else:
        time_factor = 1.00

    # SOLUTION 2
    load_factor = 1 - seats_remaining / capacity
    capacity_factor = 1 + 0.45 * load_factor
    demand_factor = 0.9 + 0.3 * route_popularity

    # SOLUTION 3
    fare = base_fare * time_factor * capacity_factor * demand_factor
    if fare < 45:
        fare = 45
    elif fare > 1500:
        fare = 1500
    return round(fare, 2)
```

### Apply the function to all five flights


```python
for flight in flights:
    # SOLUTION
    flight["fare"] = calculate_fare(
        flight["base_fare"],
        flight["days_until_departure"],
        flight["seats_remaining"],
        flight["capacity"],
        flight["route_popularity"],
    )
    print(flight["flight_id"], flight["destination"], flight["fare"])
```

**Expected fares:** 240.82, 121.92, 202.29, 199.76, 219.16.
